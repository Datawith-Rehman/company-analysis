```python
import pandas as pd 
class DataLoader:
    def __init__(self,path):
        self.path=path
        self.file=None
    def load_data(self):
        if self.path.lower().endswith('.csv'):
            self.file=pd.read_csv(self.path)
            
        elif self.path.lower().endswith('.xlsx'):
            self.file=pd.read_excel(self.path)
            
        else :
              raise ValueError('Unsupported file format')
        return self.file  
          
    def clean_col(self):
        self.file.columns=self.file.columns.str.strip()
        
    def set_AgeGroup(self):
        self.file['AgeGroup']=self.file['Customer_Age'].apply(
                                                lambda x:'Young' if x<=25 else
                                               ('Adult' if x<=35 else
                                                ('Middle_Age' if x<=55 else 'Senior')))
    def set_gender(self):
        self.file['Customer_Gender']=self.file['Customer_Gender'].replace({'M':'Male','F':'Female'})
    
    def set_date(self):
        self.file['Date']=pd.to_datetime(self.file['Date'])
        
    def drop_trash(self):
        self.file.drop(columns=['Day','Month','Age_Group'],inplace=True)    
        
     # small analysis example year wise 
     
    def yearly_profit(self) :
        self.file=self.file.set_index('Date')
        profit_summary=self.file['Profit'].resample('YE').sum().reset_index(name='total profit').sort_values(by='total profit',ascending=False)
        return profit_summary
    
    def Yearly_Revenue(self):
         self.file=self.file.set_index('Date')
         revenue=self.file['Revenue'].resample('YE').sum().reset_index(name='total revenue').sort_values(by='total revenue',ascending=False)
         return revenue
    def YearlyMargin(self):
         self.file=self.file.set_index('Date')
         profit=self.file['Profit'].resample('YE').sum().reset_index(name='total profit').sort_values(by='total profit',ascending=False)
         revenue=self.file['Revenue'].resample('YE').sum().reset_index(name='total revenue').sort_values(by='total revenue',ascending=False)
         data=profit.merge(revenue,on='Date')
         data['margin']=((data['total profit']/data['total revenue'])*100).round(2)
         return data
    


file=DataLoader(r'C:\Users\sr617\Desktop\BIKE SALES.csv')
df=file.load_data()
file.clean_col()
file.set_AgeGroup()
file.set_gender()
file.set_date()
file.Yearly_Revenue()




```
